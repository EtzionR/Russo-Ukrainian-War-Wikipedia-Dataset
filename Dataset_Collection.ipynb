## Root page (we would like to get each page connect to this page)

```python
PAGE = 'List_of_military_engagements_during_the_Russo-Ukrainian_war_(2022–present)'
PAGE
```

## "Ukraine" (only texts with this string would get into our corpus)

```python
UKR = 'Ukraine'
UKR
```

## Example of using wikipedia python api library

```python
page = wikipedia.page(wikipedia.search(PAGE)[0])
page.content
```

## simple function to get page content

```python
def get_content(title, retries=5):
    times = 0
    while times<retries:

        try:
            subpage = wikipedia.page(title, auto_suggest=False)
            return subpage.content

        except:
            try:
                subpage = wikipedia.page(wikipedia.search(title)[0], auto_suggest=False)
                return subpage.content

            except:
                content = 'None'

        finally:

            times+=1

    return content

get_content('Battle of Sumy')
```

# Collectiong data

```python
corpus = {}
errors = []

for title in tqdm(page.links):

    content = get_content(title)

    if UKR in content:
        corpus[title] = content

    sleep(1.1)

    if len(corpus)%20==0:
        with open('output.json', "w") as file:
            json.dump(corpus, file)

        print(f'\nNumber of rows: {len(corpus)}\nNumber of errors: {len(errors)}')

len(corpus)
```
